```python
import numpy as np
import pyquist as pq

f_s = 44100
N = f_s
n = np.arange(N)
samples = 0.5 * np.sin(2 * np.pi * 440 * n / f_s)    # shape (N,)

audio = pq.Audio(samples, f_s)
pq.play(audio)   # play audio out of speakers
print(audio)
# Audio(num_samples=44100, num_channels=1, sample_rate=44100)
```
