```python
# hide
import numpy as np
import pyquist as pq
f_s = 44100
n_arr = np.arange(f_s)
```

```python
sine_c = pq.Audio(np.sin(2 * np.pi * 261.63 * n_arr / f_s), f_s)
sine_e = pq.Audio(np.sin(2 * np.pi * 329.63 * n_arr / f_s), f_s)
sine_g = pq.Audio(np.sin(2 * np.pi * 392.00 * n_arr / f_s), f_s)

chord = sine_c + sine_e + sine_g
pq.play(chord / 3)
```
