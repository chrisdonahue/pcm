```python
# hide
import numpy as np
import pyquist as pq
```

```python
def amp_mod(f_c: float, f_m: float, r: float, T: float, f_s: int = 44100) -> pq.Audio:
    """Amplitude modulation: a carrier at `f_c` Hz multiplied by a modulator that
    oscillates around a nonzero average. `r` is the ratio of the carrier's amplitude
    to each sideband's amplitude (r = 2 gives the plain [1 + sin] form)."""
    t = np.arange(int(T * f_s)) / f_s
    x = np.sin(2 * np.pi * f_c * t) * (r / 2 + np.sin(2 * np.pi * f_m * t))
    return pq.Audio(x.astype(np.float32), f_s)


f_c = 220.0   # carrier frequency (Hz)
f_m = 55.0    # modulating frequency (Hz)
r = 2.0       # carrier-to-sideband amplitude ratio

# The spectrum has the carrier at f_c and two sidebands at f_c +/- f_m. Edit the
# parameters and watch them move. Can you land three sidebands on a harmonic series?
audio = amp_mod(f_c, f_m, r, T=2.0)
pq.play(audio)
pq.plot_spec(audio)
```
