```python
# hide
import numpy as np
import matplotlib.pyplot as plt
import pyquist as pq
```

```python
# The stable recursive filter y[n] = x[n] + 0.9*y[n-2]. Its impulse response is
# a train of "echoes" 2 samples apart, each 0.9x the previous one, and it decays
# forever. To implement a stable IIR filter as a finite convolution, we truncate
# its impulse response once it falls below a threshold, here -60 dB.
a, delay, N = 0.9, 2, 250
x = np.zeros(N)
x[0] = 1.0                                   # unit impulse
h = np.zeros(N)                              # the impulse response we build up
for n in range(N):
    h[n] = x[n] + a * (h[n - delay] if n >= delay else 0.0)

threshold = pq.helper.db_to_amplitude(-60.0)      # -60 dB as a linear amplitude (= 0.001)
peak_n = np.arange(0, N, delay)                   # the response peaks once per echo
cutoff = peak_n[np.argmax(np.abs(h[peak_n]) < threshold)]   # first echo below -60 dB

plt.figure(figsize=(10, 4))
ml, sl, bl = plt.stem(np.arange(N), h)            # vertical bars + dots, one per sample
plt.setp(ml, color="C4", markersize=3)
plt.setp(sl, color="C4")
plt.setp(bl, visible=False)
plt.axhline(threshold, color="C3", linestyle="--", label="-60 dB threshold (0.001)")
plt.axvline(cutoff, color="0.4", linestyle=":", label=f"truncate at n = {cutoff}")
plt.xlabel("Sample index n")
plt.ylabel("Impulse response amplitude")
plt.ylim(bottom=0)
plt.legend()
plt.show()

print(f"The impulse response decays below -60 dB at sample {cutoff}, "
      f"where we can truncate.")
```
