```python
# hide
import numpy as np
import pyquist as pq
```

```python
amen = pq.Audio.from_file("../assets/amen.wav")
f_s = amen.sample_rate
print(f"N = {len(amen):,} samples ({amen.duration:.2f} seconds at {f_s} Hz)")
pq.play(amen)
pq.plot(amen)
```

```python
beat = 0.435  # approximately the duration of one beat of the break
offset = beat * 4
duration = beat * 4

# Extract the second bar (four beats) in units of seconds...
bar = amen.segment(offset=offset, duration=duration)
pq.play(bar)

# ...or, equivalently, in units of samples
m, M = int(offset * f_s), int(duration * f_s)
same_bar = amen[m:m + M]
print(f"m = {m:,}, M = {M:,}, same samples: {np.array_equal(bar, same_bar)}")
```
