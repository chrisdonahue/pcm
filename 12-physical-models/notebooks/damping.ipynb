```python
# hide
import numpy as np
import pyquist as pq
from pyquist.helper import pitch_name_to_pitch, pitch_to_frequency
```

```python
def feedback_delay(x: np.ndarray, M: int, g: float) -> np.ndarray:
    """A delay line that loses energy on every trip: y[n] = x[n] + g * y[n - M]."""
    y = np.zeros(len(x))
    for n in range(len(x)):
        y[n] = x[n]
        if n >= M:
            y[n] += g * y[n - M]
    return y


def pluck_like(pitch: str, duration: float, f_s: int = 44100) -> pq.Audio:
    """An impulse through a damped feedback delay line that decays by 60 dB over `duration`."""
    M = round(f_s / pitch_to_frequency(pitch_name_to_pitch(pitch)))
    g = 10 ** (-3 * M / (f_s * duration))
    impulse = np.zeros(int(duration * f_s))
    impulse[0] = 1.0
    return pq.Audio(feedback_delay(impulse, M, g), f_s)


melody = pq.Score([(0.0, {"pitch": "C4", "duration": 2.0}),
                   (0.5, {"pitch": "E4", "duration": 2.0}),
                   (1.0, {"pitch": "G4", "duration": 2.0}),
                   (1.5, {"pitch": "C5", "duration": 4.0})])
pq.play(melody.render(pluck_like), normalize=True)
```
