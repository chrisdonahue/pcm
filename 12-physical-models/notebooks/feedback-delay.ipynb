```python
# hide
import numpy as np
import pyquist as pq
from pyquist.helper import pitch_name_to_pitch, pitch_to_frequency
```

```python
# collapse
def feedback_delay(x: np.ndarray, M: int) -> np.ndarray:
    """A feedback delay line, y[n] = x[n] + y[n - M], stored in a circular buffer."""
    buffer = np.zeros(M)              # the delay line: the last M output samples
    i = 0                             # a pointer to the oldest one, y[n - M]
    y = np.zeros(len(x))
    for n in range(len(x)):
        y[n] = x[n] + buffer[i]       # 1. read y[n - M]
        buffer[i] = y[n]              # 2. overwrite it with y[n]
        i = (i + 1) % M               # 3. advance the pointer, wrapping around
    return y
```

```python
f_s = 44100
M = round(f_s / 220.0)                    # tune the delay line to 220 Hz
impulse = np.zeros(f_s)                   # one second: a single 1, then zeros
impulse[0] = 1.0
h = pq.Audio(feedback_delay(impulse, M), f_s)
print(f"M = {M} audio samples, so the comb's teeth are {f_s / M:.2f} Hz apart")
pq.plot(h, duration=0.03)                                     # an impulse train...
pq.plot_freq(h, log_frequency=False).set_xlim(0, 2000)        # ...whose spectrum is a comb
pq.play(h)
```
