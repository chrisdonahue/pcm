```python
# hide
import numpy as np
import pyquist as pq
from pyquist.helper import amplitude_to_db, db_to_amplitude
```

```python
# collapse
def feedback_delay(x: np.ndarray, M: int, g: float = 1.0) -> np.ndarray:
    """A feedback delay line with attenuation, y[n] = x[n] + g * y[n - M]."""
    buffer = np.zeros(M)              # the delay line, as a circular buffer
    i = 0
    y = np.zeros(len(x))
    for n in range(len(x)):
        y[n] = x[n] + g * buffer[i]   # attenuate y[n - M] on every trip
        buffer[i] = y[n]
        i = (i + 1) % M
    return y
```

```python
#CLAUDE: Make similar changes throughout all the examples (keep input as audio. don't hardcode its sample rate)
ball = pq.Audio.from_file("../assets/ball.wav")
M = 200                                       # f_s / M = 220.5 Hz, well above 20 Hz
comb = feedback_delay(ball.samples[:, 0], M, g=0.5)
print(f"The comb's teeth are f_s / M = {ball.sample_rate / M} Hz apart")
pq.play(ball, normalize=True)  # dry
pq.play(pq.Audio(comb, ball.sample_rate), normalize=True)  # wet
```
