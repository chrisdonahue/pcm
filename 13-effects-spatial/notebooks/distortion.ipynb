```python
# hide
import numpy as np
import pyquist as pq
import matplotlib.pyplot as plt
```

```python
def hard_clip(x: np.ndarray) -> np.ndarray:
    return np.clip(x, -1.0, 1.0)


def soft_clip(x: np.ndarray) -> np.ndarray:
    return np.tanh(x)


def wavefold(x: np.ndarray) -> np.ndarray:
    return np.sin(x)                          # folds back toward zero once |x| passes pi / 2


u = np.linspace(-6, 6, 1000)
for f in [hard_clip, soft_clip, wavefold]:
    plt.plot(u, f(u), label=f.__name__)
plt.xlabel("$x$")
plt.ylabel("$f(x)$")
plt.legend()
plt.show()

guitar = pq.Audio.from_file("../assets/guitar-acoustic.wav")
drive = 24.0                                   # push the guitar far into each nonlinearity
pq.play(guitar, normalize=True)
for f in [hard_clip, soft_clip, wavefold]:
    pq.play(pq.Audio(f(drive * guitar.samples[:, 0]), guitar.sample_rate), normalize=True)
```
