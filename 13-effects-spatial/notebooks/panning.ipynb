```python
# hide
import pyquist as pq
```

```python
marimba = pq.Audio.from_file("../assets/marimba.wav").segment(offset=0.5, duration=2.0)


def panned_marimba(position: float) -> pq.Audio:
    """The marimba, panned to `position`, from -1 (left) to 1 (right)."""
    return marimba.as_stereo().pan(position, in_place=False)


score = pq.Score([(2.0 * i, {"position": p}) for i, p in enumerate([-1.0, -0.5, 0.0, 0.5, 1.0])])
pq.play(score.render(panned_marimba), normalize=True)
```
