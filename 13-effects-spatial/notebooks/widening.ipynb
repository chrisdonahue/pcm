```python
# hide
import numpy as np
import pyquist as pq
from pyquist.helper import amplitude_to_db, db_to_amplitude
```

```python
f_s = 44100
marimba = pq.Audio.from_file("../assets/marimba.wav").samples[:, 0]
D = round(0.012 * f_s)                                  # a 12 ms delay
mid = 2 * marimba                                       # x_L + x_R, both equal to the mono marimba
side = np.concatenate([np.zeros(D), marimba[:-D]])      # an invented side signal: a delayed copy
for width in [0.0, 0.5, 1.0]:
    left, right = (mid + width * side) / 2, (mid - width * side) / 2
    pq.play(pq.Audio(np.stack([left, right], axis=1), f_s), normalize=True)
```
