```python
# hide
import numpy as np
import pyquist as pq
```

```python
pad = pq.Audio.from_file("../assets/pad-stereo.wav")
left, right = pad.samples[:, 0], pad.samples[:, 1]
mid, side = left + right, left - right
for width in [0.0, 1.0, 2.0]:                 # mono, unchanged, and wider
    widened = np.stack([(mid + width * side) / 2, (mid - width * side) / 2], axis=1)
    pq.play(pq.Audio(widened, pad.sample_rate), normalize=True)
```
