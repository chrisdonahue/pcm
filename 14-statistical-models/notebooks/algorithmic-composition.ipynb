```python
# hide
import numpy as np
import pyquist as pq
```

```python
# hide
def tone(pitch: float, duration: float, f_s: int = 44100, **kwargs) -> pq.Audio:
    """A sinusoid with a quick attack and an exponential decay, like a small bell."""
    t = np.arange(int(duration * f_s)) / f_s
    envelope = np.minimum(t / 0.005, 1.0) * np.exp(-4 * t / duration)
    return pq.Audio(envelope * np.sin(2 * np.pi * pq.helper.pitch_to_frequency(pitch) * t), f_s)
```

```python
def tintinnabuli(melody: list[int], chord: list[int], ioi: float = 0.5) -> pq.Score:
    """Accompany each note of the melody with the nearest note of the chord above it."""
    events = []
    for i, pitch in enumerate(melody):
        above = pitch + 1
        while above % 12 not in chord:          # step up until we reach a note of the chord
            above += 1
        events.append((i * ioi, {"pitch": pitch, "duration": 3 * ioi}))
        events.append((i * ioi, {"pitch": above, "duration": 3 * ioi}))
    return pq.Score(events)


a_minor = [9, 0, 4]                             # the pitch classes of A, C, and E
melody = [69, 71, 72, 74, 72, 71, 69, 67, 65, 67, 69, 71, 69]
score = tintinnabuli(melody, a_minor)
pq.play(score.render(tone), normalize=True)
```
