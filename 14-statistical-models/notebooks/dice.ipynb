```python
alphabet = [1, 2, 3, 4, 5, 6]
weights = np.ones(6)
weights /= weights.sum()                        # a fair die
print(np.random.choice(alphabet, size=10, p=weights))

rolls = np.random.choice(alphabet, size=6000, p=weights)
for face in alphabet:
    print(f"{face}: {np.sum(rolls == face)} rolls")
```
