```python
# hide
# Approximate frequency (Hz) and level (dB SPL) of some familiar sounds.
SOUNDS = [
    ("🍃", "Rustling leaves", 3000, 15),
    ("🦟", "Mosquito", 600, 25),
    ("🤫", "Whisper", 2000, 30),
    ("🐦", "Birdsong", 5000, 45),
    ("💧", "Dripping faucet", 3500, 35),
    ("🧊", "Refrigerator hum", 125, 40),
    ("🗣️", "Conversation", 700, 60),
    ("🐈", "Cat meowing", 1000, 70),
    ("🎹", "Piano", 400, 75),
    ("☎️", "Telephone ringing", 2500, 80),
    ("🐕", "Dog barking", 800, 85),
    ("🚚", "Truck", 160, 90),
    ("🏍️", "Motorcycle", 250, 100),
    ("🪚", "Chainsaw", 1500, 110),
    ("🌩️", "Thunder", 60, 120),
    ("✈️", "Jet engine", 1000, 130),
]
OCTAVES = [62.5, 125, 250, 500, 1000, 2000, 4000, 8000]


def figure():
    fig = go.Figure()
    fig.add_scatter(x=[40, 12000], y=[0, 0], mode="lines+text", line=dict(color=STEEL, dash="dash"),
                    text=["", "threshold of hearing"], textposition="top left",
                    textfont=dict(color=IRON), hoverinfo="skip")
    fig.add_scatter(x=[40, 12000], y=[120, 120], mode="lines+text",
                    line=dict(color=RED, dash="dash"), text=["", "threshold of pain"],
                    textposition="bottom left", textfont=dict(color=RED), hoverinfo="skip")
    fig.add_scatter(
        x=[f for _, _, f, _ in SOUNDS], y=[db for _, _, _, db in SOUNDS],
        mode="text", text=[e for e, _, _, _ in SOUNDS], textfont=dict(size=26),
        customdata=[name for _, name, _, _ in SOUNDS],
        hovertemplate="%{customdata}<br>~%{x} Hz, ~%{y} dB SPL<extra></extra>")
    fig.update_xaxes(type="log", range=[np.log10(40), np.log10(12000)], tickvals=OCTAVES,
                     ticktext=["62.5", "125", "250", "500", "1k", "2k", "4k", "8k"],
                     title_text="Frequency (Hz, log scale)", fixedrange=True)
    fig.update_yaxes(range=[-10, 140], dtick=20, title_text="Sound pressure level (dB SPL)",
                     fixedrange=True)
    fig.update_layout(height=460)
    return fig


icm_plotly.show(figure)
```
