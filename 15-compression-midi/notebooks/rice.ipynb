```python
def zigzag(v: int) -> int:
    """Interleave signed integers as 0, -1, 1, -2, 2, ... -> 0, 1, 2, 3, 4, ..."""
    return 2 * v if v >= 0 else -2 * v - 1


def rice_encode(v: int, k: int) -> str:
    """The Rice code of a signed integer v with parameter k, as a string of bits."""
    u = zigzag(v)
    q, r = u >> k, u & ((1 << k) - 1)        # quotient and remainder of u / 2^k
    unary = "1" * q + "0"                      # q ones, then a terminating zero
    binary = format(r, f"0{k}b") if k else ""  # the remainder, in exactly k bits
    return unary + binary


for v in [0, -1, 1, 5, -60]:
    code = rice_encode(v, k=4)
    print(f"{v:4d} -> {code} ({len(code)} bits)")
```
